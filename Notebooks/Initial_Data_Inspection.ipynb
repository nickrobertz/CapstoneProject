# Initial Data Inspection of Black-Box Functions

This notebook provides a simple inspection of the initial data for all 8 black-box optimization functions.

## Overview

We have 8 different black-box functions with varying characteristics:
- Function 1: Contamination detection (2D, 10 samples, sparse signal)
- Function 2: Noisy log-likelihood (2D, 10 samples, local optima)
- Function 3: Drug discovery (3D, 15 samples, minimize side effects)
- Function 4: Warehouse optimization (4D, 30 samples, dynamic)
- Function 5: Chemical process yield (4D, 20 samples, unimodal)
- Function 6: Cake recipe optimization (5D, 20 samples, negative scores)
- Function 7: ML hyperparameter tuning (6D, 30 samples)
- Function 8: High-dimensional optimization (8D, 40 samples)


```python
import numpy as np
import pandas as pd
from pathlib import Path
```

## 1. Load All Data

Load all the initial data from the 8 functions and display summary statistics.


```python
# Define data path
data_path = Path('../Data/InitialData')

# Load all functions data
functions_data = {}

for i in range(1, 9):
    func_name = f'function_{i}'
    func_path = data_path / func_name
    
    inputs = np.load(func_path / 'initial_inputs.npy')
    outputs = np.load(func_path / 'initial_outputs.npy')
    
    functions_data[func_name] = {
        'inputs': inputs,
        'outputs': outputs,
        'n_samples': inputs.shape[0],
        'n_dims': inputs.shape[1]
    }
    
    print(f"{func_name}: {inputs.shape[0]} samples, {inputs.shape[1]}D input")
    print(f"  Input range: [{inputs.min():.3f}, {inputs.max():.3f}]")
    print(f"  Output range: [{outputs.min():.3f}, {outputs.max():.3f}]")
    print()
```

## 2. Summary Statistics Table


```python
# Create summary statistics table
summary_data = []
for func_name, data in functions_data.items():
    outputs = data['outputs']
    summary_data.append({
        'Function': func_name.replace('_', ' ').title(),
        'Dimensions': data['n_dims'],
        'Samples': data['n_samples'],
        'Output Mean': f"{outputs.mean():.3f}",
        'Output Std': f"{outputs.std():.3f}",
        'Output Min': f"{outputs.min():.3f}",
        'Output Max': f"{outputs.max():.3f}",
        'Output Range': f"{outputs.max() - outputs.min():.3f}"
    })

summary_df = pd.DataFrame(summary_data)
print("\n" + "="*100)
print("SUMMARY OF ALL BLACK-BOX FUNCTIONS")
print("="*100)
print(summary_df.to_string(index=False))
print("="*100)
```

## 3. Key Insights and Observations

### Function Characteristics:

1. Function 1 (Contamination Detection): 
   - Sparse signal with mostly zero outputs - requires extensive exploration
   - 2D problem with only 10 samples
   - Need methods that can detect multiple separate peaks

2. Function 2 (Noisy Log-Likelihood): 
   - Continuous but noisy landscape with local optima
   - 2D problem, easier than F1 due to continuous nature
   - Balance exploration and exploitation

3. Function 3 (Drug Discovery): 
   - 3D problem with moderate sampling (15 samples)
   - Negative optimization (minimizing side effects)
   - Need conservative approach due to experimental noise

4. Function 4 (Warehouse Optimization): 
   - 4D dynamic problem with largest sample size (30)
   - Full of local optima
   - Emphasize robustness and repeatability

5. Function 5 (Chemical Process Yield): 
   - Unimodal 4D - easiest to optimize
   - Single peak means more exploitation possible
   - 20 samples available

6. Function 6 (Cake Recipe Optimization): 
   - 5D with all negative scores
   - Goal is to approach zero (maximize negative of sum)
   - Human-evaluated scores introduce noise

7. Function 7 (ML Hyperparameter Tuning): 
   - 6D standard hyperparameter tuning
   - 30 samples available
   - Can leverage prior knowledge of good ranges

8. Function 8 (High-Dimensional Optimization): 
   - Most challenging 8D problem
   - Largest sample size (40) but highest dimensionality
   - Finding good local optima may be practical goal

### Optimization Strategies:

- Low-dimensional (F1-F2): Can visualize directly, focus on exploration
- Medium-dimensional (F3-F5): Balance exploration/exploitation
- High-dimensional (F6-F8): Need specialized strategies, dimension reduction techniques
- Sparse signals (F1): Require aggressive exploration
- Unimodal (F5): Can be more exploitative
- Noisy (F2, F6): Need robust methods that handle noise
- Dynamic (F4): Emphasize validation and robustness
